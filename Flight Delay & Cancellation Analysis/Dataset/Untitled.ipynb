```python
import pandas as pd
```

```python
file_path = r"C:\Users\rajkumar\Downloads\Airlines+Airports+Cancellation+Codes+&+Flights\flights.csv"

df = pd.read_csv(
    file_path,
    dtype={
        "ORIGIN_AIRPORT": "string",
        "DESTINATION_AIRPORT": "string"
    }
)
```

```python
################ converting to TIME
```

```python
time_cols = [
    'SCHEDULED_DEPARTURE',
    'DEPARTURE_TIME',
    'WHEELS_OFF',
    'WHEELS_ON',
    'SCHEDULED_ARRIVAL',
    'ARRIVAL_TIME'
]

# Convert all time columns into HH:MM format
for col in time_cols:
    df[col] = pd.to_datetime(
        df[col].astype('Int64').astype('string').str.zfill(4),
        format='%H%M',
        errors='coerce'
    ).dt.time

print(df[time_cols].head())
```

```python
import os

folder_path = r"C:\Users\rajkumar\Downloads\Airlines+Airports+Cancellation+Codes+&+Flights"

if os.path.exists(folder_path):
    file_path = os.path.join(folder_path, "flights.parquet")
    df.to_parquet(file_path, engine="pyarrow", index=False)
    print("File saved successfully:", file_path)

else:
    print("Folder does not exist!")
```
